# **PART A: Loading and Configuring a Pre-trained LLM**

```python

# Name: Lathika M
# Register Number: 24BAD062
# Experiment No: 9 - Part A

!pip install transformers torch

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load pre-trained model
model_name = "microsoft/DialoGPT-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Prepare user prompt
prompt = "Hello, how are you?"

# Tokenization
inputs = tokenizer.encode(
    prompt + tokenizer.eos_token,
    return_tensors="pt"
)

# Generate response
outputs = model.generate(
    inputs,
    max_new_tokens=50,
    temperature=0.7,
    do_sample=True,
    pad_token_id=tokenizer.eos_token_id
)

# Decode output
response = tokenizer.decode(
    outputs[:, inputs.shape[-1]:][0],
    skip_special_tokens=True
)

print("Name: Lathika M")
print("Register Number: 24BAD062")
print("User:", prompt)
print("Chatbot:", response)
```

# **PART B: IMPLEMENTING CHATBOT FUNCTIONALITY**

```python

# Name: Lathika M
# Register Number: 24BAD062
# Experiment No: 9 - Part B

from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "microsoft/DialoGPT-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

chat_history = None

def chatbot(user_input):
    global chat_history

    # Convert user input into tokens
    new_input = tokenizer.encode(
        user_input + tokenizer.eos_token,
        return_tensors="pt"
    )

    # Add previous conversation
    if chat_history is not None:
        bot_input = torch.cat(
            [chat_history, new_input], dim=-1
        )
    else:
        bot_input = new_input

    # Generate response
    chat_history = model.generate(
        bot_input,
        max_new_tokens=50,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    # Decode response
    response = tokenizer.decode(
        chat_history[:, bot_input.shape[-1]:][0],
        skip_special_tokens=True
    )

    return response

import torch

# Test chatbot
print("Name: Lathika M")
print("Register Number: 24BAD062")

print("User: Hello")
print("Chatbot:", chatbot("Hello"))

print("User: What is AI?")
print("Chatbot:", chatbot("What is AI?"))
```
